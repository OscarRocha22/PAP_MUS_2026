```python
def cleaning(df,year):

    clean_df=df[df['error'].isna()]
    return clean_df.to_csv(f"./data/clean/clean_{year}.csv")

    
dfs= [news2020, news2021, news2022, news2023, news2024, news2025, news2026]
year=2020
for df in dfs:
    cleaning(df,year)
    year+=1
```
